```python
import numpy as np
def sigmoid(x):
    """
    Calculate sigmoid
    """
    return 1/(1+np.exp(-x))
def sigmoid_prime(x):
    """
    # Derivative of the sigmoid function
    """
    return sigmoid(x) * (1 - sigmoid(x))
learnrate = 0.5
x = np.array([1, 2, 3, 4])
y = np.array(0.5)

# Initial weights
w = np.array([0.5, -0.5, 0.3, 0.1])

### Calculate one gradient descent step for each weight
### Note: Some steps have been consolidated, so there are
###       fewer variable names than in the above sample code
# TODO: Calculate the node's linear combination of inputs and weights
h = np.dot(x, w)

### Notebook grading
import numpy as np
x_test = np.array([1, 2, 3, 4])
y_test = np.array(0.5)
w_test = np.array([0.5, -0.5, 0.3, 0.1])
learnrate_test = 0.5

h_answer = np.dot(x_test, w_test)
if h == h_answer:
    print("Good job!")
else:
    print("Try again. `h` is not correct.")
```

```python
# TODO: Calculate output of neural network
nn_output = sigmoid(h)

### Notebook grading
nn_output_answer = 1/(1+np.exp(-np.dot(x_test, w_test)))

if nn_output == nn_output_answer:
    print("Good job!")
else:
    print("Try again. `nn_output` is not correct.")
```

```python
# TODO: Calculate error of neural network
error = y - nn_output

### Notebook grading
error_answer = y_test - nn_output_answer

if error == error_answer:
    print("Good job!")
else:
    print("Try again. `error` is not correct.")
```

```python
# TODO: Calculate the error term
#       Remember, this requires the output gradient, which we haven't
#       specifically added a variable for.
error_term = error * nn_output * (1 - nn_output)

### Notebook grading
error_term_answer = error_answer * nn_output_answer * (1 - nn_output_answer)

if error_term == error_term_answer:
    print("Good job!")
else:
    print("Try again. `error_term` is not correct.")
```

```python
# TODO: Calculate change in weights
del_w = learnrate * error_term * x

# Optional print
print('Neural Network output:')
print(nn_output)
print('Amount of Error:')
print(error)
print('Change in Weights:')
print(del_w)

### Notebook grading
del_w_answer = learnrate_test * error_answer * nn_output_answer * (1 - nn_output_answer) * x_test

test = del_w == del_w_answer
if test.all():
    print("Good job!")
else:
    print("Try again. `del_w` is not correct.")
```

```python
import numpy as np
from data_prep import features, targets, features_test, targets_test
from helper import sigmoid, grade_implement_gd

np.random.seed(42)

n_records, n_features = features.shape
last_loss = None

# Initialize weights
weights = np.random.normal(scale=1 / n_features ** 0.5, size=n_features)

# Hyperparameters
epochs = 1000
learnrate = 0.5
```

```python
for e in range(epochs):
    del_w = np.zeros(weights.shape)
    for x, y in zip(features.values, targets):

        # TODO: Calculate the output
        #       Hint: sigmoid of the dot product of x and weights
        output = sigmoid(np.dot(x, weights))

        # TODO: Calculate the error  (y - output)
        error = y - output

        # TODO: Calculate the error term
        #       Hint: error * output * (1 - output)  — sigmoid derivative
        error_term = error * output * (1 - output)

        # TODO: Accumulate the change in weights for this sample
        #       Hint: error_term * x
        del_w += error_term * x

    # TODO: Update weights using the learning rate and the average change
    #       Hint: learnrate * del_w / n_records
    weights += learnrate * del_w / n_records

    # Print training loss every 10% of epochs
    if e % (epochs / 10) == 0:
        out  = sigmoid(np.dot(features, weights))
        loss = np.mean((out - targets) ** 2)
        if last_loss and last_loss < loss:
            print("Train loss: ", loss, "  WARNING - Loss Increasing")
        else:
            print("Train loss: ", loss)
        last_loss = loss

# ── Grading ──────────────────────────────────────────────────────────────────
grade_implement_gd(weights, features_test, targets_test)
```

```python
import numpy as np
from helper import (
    sigmoid,
    grade_output_error,
    grade_output_error_term,
    grade_hidden_error_term,
    grade_delta_w_h_o,
    grade_delta_w_i_h
)

x = np.array([0.5, 0.1, -0.2])
target = 0.6
learnrate = 0.5

weights_input_hidden = np.array([[0.5, -0.6],
                                 [0.1, -0.2],
                                 [0.1,  0.7]])
weights_hidden_output = np.array([0.1, -0.3])

## Forward pass
hidden_layer_input  = np.dot(x, weights_input_hidden)
hidden_layer_output = sigmoid(hidden_layer_input)
output_layer_in     = np.dot(hidden_layer_output, weights_hidden_output)
output              = sigmoid(output_layer_in)
```

```python
## Backwards pass

# TODO: Calculate output error
error = target - output

grade_output_error(error, target, output)
```

```python
# TODO: Calculate error term for output layer
output_error_term = error * output * (1 - output)

grade_output_error_term(output_error_term, error, output)
```

```python
# TODO: Calculate error term for hidden layer
hidden_error_term = (np.dot(output_error_term, weights_hidden_output)
                     * hidden_layer_output
                     * (1 - hidden_layer_output))

grade_hidden_error_term(hidden_error_term, output_error_term, weights_hidden_output, hidden_layer_output)
```

```python
# TODO: Calculate change in weights for hidden layer to output layer
delta_w_h_o = learnrate * output_error_term * hidden_layer_output

grade_delta_w_h_o(delta_w_h_o, learnrate, output_error_term, hidden_layer_output)
```

```python
# TODO: Calculate change in weights for input layer to hidden layer
delta_w_i_h = learnrate * hidden_error_term * x[:, None]

print('Change in weights for hidden layer to output layer:')
print(delta_w_h_o)
print('Change in weights for input layer to hidden layer:')
print(delta_w_i_h)

grade_delta_w_i_h(delta_w_i_h, learnrate, hidden_error_term, x)
```
