# Transformers for Sentiment Analysis

You may have seen transformer models being used for a generation task. In this project you will build on your generation-style Transformer knowledge to have the transformer perform a binary classification task. This task requires a few differences in approach compared to a generation task. We'll explore those differences below.

## 1. Data

When we do a generation task, we look at a continuous text stream and split it into overlapping segments to guess the next token. For sentiment classification, each movie review is a single data point paired with a **label** (positive or negative). There's no sliding window. Instead, we have a neat table of reviews and their sentiments. That's the core difference: in generation, you focus on predicting the next token in a sequence; in classification, you focus on predicting a label for the entire sequence.

## 2. Tokenization

Previously, you may have seen character-level tokenization, where every single character is treated as a token. That's straightforward but can lead to very long input sequences and doesn't scale well. In this project, we use **subword tokenization** (BERT-based-uncased). Instead of splitting strictly at each character, the tokenizer identifies common subword units (like `trans` and `##formers`), giving a better balance between efficiency and coverage of different words.

## 3. Output Layer and Model Focus

In a generation model, the Transformer typically outputs a set of token-level predictions, and you grab the most recent token's output to decide the *next* token in the generated text. Here in classification, we condense all token embeddings into a single pooled representation (for example, by taking the average). We then feed that single vector into a classification head that outputs something like two **logits**—one for **positive**, one for **negative**. We're no longer generating text token by token; we're simply classifying whether a whole review is positive or negative.

## 4. Training with epochs

You've also seen how a dataset can be passed through a `RandomShuffler` pipeline for generation tasks, where tokens are shuffled or sampled in a particular way across many training steps. Here, we're simplifying and adopting the more common **epochs** approach. Each epoch is one complete pass over the entire dataset, and we pass through the entire data multiple times. We still shuffle (to avoid memorizing the order of samples), but now it's typically done automatically in the `DataLoader`, and we do a fixed number of full passes (epochs) through the dataset. This clarifies how often the model has "seen" every review during training and helps us measure progress between epochs.
