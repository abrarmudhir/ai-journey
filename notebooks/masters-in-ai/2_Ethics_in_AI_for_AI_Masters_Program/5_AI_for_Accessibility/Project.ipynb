# Exercise: Consumer Perspective

## Scenario

Suppose a new AI-based voice assistant technology, **HelperBot**, is coming out from XYZ Company. HelperBot is an AI-powered voice assistant that allows people to engage with it using speech and responds back to the user in speech. It can support tasks such as:

- Answering short internet queries, such as *What is the weather in Seattle?*
- Integrating with smart-home devices in the space
- Integrating with other apps, such as *Put lightbulbs in my shopping cart*
- Playing short games such as 20 Questions

HelperBot also has potential accessibility applications:

- It is an accessible way to query the internet or issue smart-home commands for people with limited or no typing dexterity.
- It can respond to commands with limited mental context switching from the user, which may benefit some neurodivergent people.

However, people at XYZ Company want the tool to be built inclusively, and end users of all abilities should ideally be able to use it.

Imagine that **you are someone who is thinking of buying HelperBot**. Consider buying it for a space where people have a variety of disabilities, including someone with a motor disability, someone who is blind, and someone who is Deaf. For example, it might be used in a community center or a team's office.

## Instructions

Answer each section below from the consumer's perspective. Consider the needs of people with different disabilities and identify potential risks, exclusions, and mitigations.

## Dataset

*What questions would you have about the dataset that the model was trained on?*

**Answer:**

- What datasets was it trained on?
- What disabilities were represented in this dataset?
  - Were there people with dyspraxia in the dataset?
  - Were there people with Deaf accents in the dataset?
  - Were there people who use assistive technologies that might make noise that is audible in voice recordings?
  - Were there people who had multiple marginalized identities, such as someone who is Deaf and has an accent from a language that is less well recognized by typical AI systems?
- Was the data collected ethically?
  - Did people know that their data was going to be used for this purpose?
  - Were they compensated appropriately?

## Outputs

Let's pretend that the dataset was inclusive, represented people with disabilities well, and that the system could correctly interpret the speech of people with different types of speech.

Can everyone on your team or in your group use the system effectively? Who might be left out of this design?

**Answer:**

Without further system design, the following groups might have a hard time using the system:

- People who communicate nonverbally might not be able to issue commands to the system. This could include:
  - People who are d/Deaf or hard of hearing and use a sign language
  - People who have cognitive impairments and prefer to communicate nonverbally using pictures and images
  - People who have late-stage ALS and use text-to-speech tools or alternative, augmentative communication devices
- People who speak quietly and might not be picked up by the microphones
- People who speak slowly and with breaks that cause the system to think they are done talking before they are
- People who have trouble hearing, including people who are d/Deaf, hard of hearing, or experiencing hearing loss

## Impact

*What would be the impact of the inaccessibility if people could not use the system, or if it mis-interpreted their voice?*

**Answer:**

- People will not be able to complete the task they set out to do and feel sad or frustrated.
- People may feel frustrated that they are experiencing bias or having a marginalized identity.
- People may feel left out if a group is performing an activity with the device that they cannot participate in.

# Exercise: Developer Perspective

## Scenario

Suppose a new AI-based voice assistant technology, **HelperBot**, is coming out from XYZ Company. HelperBot is an AI-powered voice assistant that allows people to engage with it using speech and responds back to the user in speech. It can support tasks such as:

- Answering short internet queries, such as *What is the weather in Seattle?*
- Integrating with smart-home devices in the space
- Integrating with other apps, such as *Put lightbulbs in my shopping cart*
- Playing short games such as 20 Questions

HelperBot has potential accessibility applications for people with limited or no typing dexterity and for some neurodivergent people. The development team wants the tool to be built inclusively for users with a variety of disabilities, including motor disabilities, blindness, and Deafness.

Imagine that **you are a member of the team creating HelperBot**. You want to build the product so that it can be used by people with a variety of disabilities.

## Dataset

*What data would you want to train this model on? What traits or scenarios would you want to be sure are represented in your data?*

**Answer:**

- Look for publicly available datasets and check whether people with disabilities are represented, and how much they are represented. For example, if they are only one sample in a dataset of tens of thousands, that is likely not enough.
- Explore whether people with different disabilities are represented, including:
  - People with dyspraxia
  - People with Deaf accents
  - People who use assistive technologies that might make noise audible in voice recordings
  - People with multiple marginalized identities, such as someone who is Deaf and has an accent from a language that is less well represented in the dataset

## Data Collection

*Let's say that this data is not already readily available and needs to be collected. What would you do to make sure that the data-collecting process is fair and ethical?*

**Answer:**

- Ensure that people know where and how their data is going to be used.
- Compensate people for their data.
- If people are required to record and submit new data, consider the burden on them, such as whether the collection procedure is overly fatiguing or highly time-consuming.

## Outputs

Let's pretend that the dataset was inclusive and represented people with disabilities well, and that the system can correctly interpret the speech of people with all different types of speech.

Who might still be left out of this design?

**Answer:**

- Without further system design, the following groups might have a hard time using the system:
  - People who communicate nonverbally might not be able to issue commands to the system, including:
    - People who are d/Deaf or hard of hearing and use a sign language and/or written English
    - People who have cognitive impairments and prefer to communicate nonverbally using pictures and images
    - People who have late-stage ALS and use text-to-speech tools or alternative, augmentative communication devices
  - People who speak quietly and might not be picked up by the microphones
  - People who speak slowly and with breaks that cause the system to think they are done talking before they are
  - People who have trouble hearing, including people who are d/Deaf, hard of hearing, or experiencing hearing loss

To better include these groups, consider multimodal input and output, such as allowing typed text as input and output, and making the length of time the system waits for user input adjustable.

## Impact

*What would be the impact of the inaccessibility for people if they could not use the system, or if it mis-interpreted their voice?*

**Answer:**

- People will not be able to complete the task they set out to do and may feel sad or frustrated.
- People may feel frustrated that they are experiencing bias or having a marginalized identity.
- People may feel left out if a group is performing an activity with the device that they cannot participate in.
