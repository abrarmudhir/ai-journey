# Project Overview

## Project Introduction

Personalization is a central aspect of many core AI systems. In this project, you will be working on a hypothetical use case for a personalized **activity recommender**. The use case has a medium ethical AI risk level and involves a synthetic tabular dataset.

IDOOU is a mobile app users can leverage to get recommendations on activities they can take in a given area, like "visiting a movie theater," "visiting a park," "sightseeing," "hiking," or "visiting a library."

IDOOU's differentiating value proposition is two-fold:

1. The app offers **personalization**, using features such as gender, age, and education level to predict users' interests and the right type of recommendations.
2. The app's objective is to remove users from having to handle the nitty-gritty details of finding the right activity, like determining the appropriate budget, ensuring the weather is perfect, and the location/accommodation is not closed. This way, users can focus on what matters: having fun!

The engineering team behind the app has designed IDOOU to be fairly flexible and ambitious in the use cases the app can support. Hotels can recommend users install IDOOU to act as a smart concierge-type of application, and IDOOU can be integrated as part of autonomous vehicles' dashboards to recommend local locations users can visit while driving around town.

**Problem statement:**

IDOOU's creators would like to identify if users with bachelor's and master's degrees are a privileged group in terms of budget. In other words, do users with higher education credentials beyond high school have a budget >= $300 compared to users of the app who graduated from high school?

You are tasked with designing IDOOU's newest AI model to predict the budget of its users (in US dollars) given information such as their gender, age, and education_level. You will also explore the provided data and analyze and evaluate this budget predictor's fairness and bias.

We will be relying on IBM's AIF360 Toolkit for this project.

## Key points

- The training data was conducted through a user experience study of about 300,000 participants.
- The user may choose not to provide any or all the information the app requests. The training data also reflects this.
- Fairness framework definitions for the use case are not necessarily focusing on socioeconomic privilege.

## Step 1: Data Pre-Processing and Evaluation

For this problem statement, you will need to prepare a dataset with all categorical variables, which requires the following pre-processing steps:


- Remove the NA values from the dataset
- Convert Age and Budget (in dollars) to categorical columns with the following binning:

> Bins for Age: 18-24, 25-44, 45-65, 66-92

> Bins for Budget: >=300, <300

```python
act_rec_dataset = act_rec_dataset_raw.dropna().copy()
act_rec_dataset['Age'] = pd.cut(act_rec_dataset['Age'], bins=[17,24,44,65,92], labels=['18-24','25-44','45-65','66-92'])
act_rec_dataset['Budget (in dollars)'] = pd.cut(act_rec_dataset['Budget (in dollars)'], bins=[-1,299,np.inf], labels=['<300','>=300'])
print(f"Pre-processed dataset shape: {act_rec_dataset.shape}")
print(f"Remaining missing values: {int(act_rec_dataset.isna().sum().sum())}")
act_rec_dataset.head()
```

### Evaluate bias issues in the dataset

Next, let's take a look at potential hints of data bias in the variables, particularly the "Gender", "Age", and "Education" variables.

Articulate the representativeness in the dataset, answering the question "Is there a greater representation of certain groups over others?"

```python
fig, axes = plt.subplots(1,3,figsize=(18,5))
for ax, col, title in zip(axes,['Gender','Age','Education_Level'],['Representation by Gender','Representation by Age Group','Representation by Education Level']):
    order=act_rec_dataset[col].value_counts().index
    sns.countplot(data=act_rec_dataset,x=col,order=order,ax=ax,color='#4472C4')
    ax.set_title(title); ax.set_xlabel(col.replace('_',' ')); ax.set_ylabel('Number of records'); ax.tick_params(axis='x',rotation=35)
fig.tight_layout(); fig.savefig('images/01_group_representation.png',dpi=160,bbox_inches='tight'); plt.show()
```

**Your answer:** The dataset is not evenly representative. Age is concentrated in the youngest bands, and the education groups are not equally sized after missing records are removed. Gender categories are comparatively similar in size, but missing values were excluded and therefore are not represented in the analysis. The strong imbalance in the outcome by education level is more consequential than modest differences in group counts: users with Bachelor's or Master's degrees are much more likely to have a budget of at least $300 than high-school graduates.

For this project, we are using a train-test-validation split.

You have available boilerplate for training 2 ML models on this dataset - you will need to train these models and use the methods we covered in this course to identify and evaluate their performance (**using the accuracy metric and confusion matrix**).

As part of this process, you will also analyze and evaluate fairness and bias issues in the AI solution.

```python
indices=np.arange(len(act_rec_dataset))
train_idx,remainder_idx=train_test_split(indices,test_size=0.5,random_state=RANDOM_STATE,stratify=binary_act_dataset.labels.ravel())
validate_idx,test_idx=train_test_split(remainder_idx,test_size=0.4,random_state=RANDOM_STATE,stratify=binary_act_dataset.labels.ravel()[remainder_idx])
orig_train=binary_act_dataset.subset(train_idx); orig_validate=binary_act_dataset.subset(validate_idx); orig_test=binary_act_dataset.subset(test_idx)
print(f"Train/validation/test sizes: {len(orig_train.features):,}/{len(orig_validate.features):,}/{len(orig_test.features):,}")
```

```python
def make_prediction_dataset(dataset,predictions):
    result=dataset.copy(deepcopy=True); result.labels=np.asarray(predictions,dtype=float).reshape(-1,1); return result

def fairness_metrics(dataset,predictions):
    metric=ClassificationMetric(dataset,make_prediction_dataset(dataset,predictions),unprivileged_groups=unprivileged_groups,privileged_groups=privileged_groups)
    return {'balanced_accuracy':(metric.true_positive_rate()+metric.true_negative_rate())/2,'average_odds_difference':metric.average_odds_difference(),'statistical_parity_difference':metric.statistical_parity_difference(),'equal_opportunity_difference':metric.equal_opportunity_difference(),'theil_index':metric.theil_index(),'disparate_impact':metric.disparate_impact()}

def choose_threshold(model,dataset,thresholds=np.linspace(.01,.99,99)):
    p=model.predict_proba(dataset.features)[:,1]; scores=[balanced_accuracy_score(dataset.labels.ravel(),(p>=t).astype(int)) for t in thresholds]; j=int(np.argmax(scores)); return float(thresholds[j]),dict(zip(thresholds,scores))

def describe_metrics(metrics,threshold):
    print(f"Threshold: {threshold:.4f}")
    for k,v in metrics.items(): print(f"{k}: {v:.6f}")
```

```python
GNB_model=GaussianNB().fit(orig_train.features,orig_train.labels.ravel(),orig_train.instance_weights)
gnb_threshold,gnb_validation_scores=choose_threshold(GNB_model,orig_validate)
gnb_pred=(GNB_model.predict_proba(orig_test.features)[:,1]>=gnb_threshold).astype(int)
gnb_metrics=fairness_metrics(orig_test,gnb_pred)
print('Gaussian Naive Bayes test metrics:'); describe_metrics(gnb_metrics,gnb_threshold)
```

```python
gnb_accuracy=accuracy_score(orig_test.labels.ravel(),gnb_pred); gnb_balanced_accuracy=balanced_accuracy_score(orig_test.labels.ravel(),gnb_pred)
print(f"Gaussian Naive Bayes accuracy: {gnb_accuracy:.4f}"); print(f"Gaussian Naive Bayes balanced accuracy: {gnb_balanced_accuracy:.4f}")
fig,ax=plt.subplots(figsize=(5,4)); ConfusionMatrixDisplay.from_predictions(orig_test.labels.ravel(),gnb_pred,display_labels=['Budget < $300','Budget >= $300'],cmap='Blues',ax=ax,colorbar=False); ax.set_title('Gaussian Naive Bayes Confusion Matrix'); fig.tight_layout(); fig.savefig('images/03_gnb_confusion_matrix.png',dpi=160,bbox_inches='tight'); plt.show()
```

```python
LR_model=make_pipeline(StandardScaler(),LogisticRegression(max_iter=500,random_state=RANDOM_STATE))
LR_model.fit(orig_train.features,orig_train.labels.ravel(),logisticregression__sample_weight=orig_train.instance_weights)
```

```python
lr_threshold,lr_validation_scores=choose_threshold(LR_model,orig_validate)
lr_pred=(LR_model.predict_proba(orig_test.features)[:,1]>=lr_threshold).astype(int); lr_metrics=fairness_metrics(orig_test,lr_pred)
print('Logistic Regression test metrics:'); describe_metrics(lr_metrics,lr_threshold)
```

```python
lr_accuracy=accuracy_score(orig_test.labels.ravel(),lr_pred); lr_balanced_accuracy=balanced_accuracy_score(orig_test.labels.ravel(),lr_pred)
print(f"Logistic Regression accuracy: {lr_accuracy:.4f}"); print(f"Logistic Regression balanced accuracy: {lr_balanced_accuracy:.4f}")
fig,ax=plt.subplots(figsize=(5,4)); ConfusionMatrixDisplay.from_predictions(orig_test.labels.ravel(),lr_pred,display_labels=['Budget < $300','Budget >= $300'],cmap='Greens',ax=ax,colorbar=False); ax.set_title('Logistic Regression Confusion Matrix'); fig.tight_layout(); fig.savefig('images/04_lr_confusion_matrix.png',dpi=160,bbox_inches='tight'); plt.show()
```

```python
accuracy_table=pd.DataFrame({'Model':['Gaussian Naive Bayes','Logistic Regression'],'Accuracy':[gnb_accuracy,lr_accuracy],'Balanced Accuracy':[gnb_balanced_accuracy,lr_balanced_accuracy]})
print(accuracy_table.to_string(index=False)); fig,ax=plt.subplots(figsize=(7,4)); accuracy_table.set_index('Model')[['Accuracy','Balanced Accuracy']].plot(kind='bar',ax=ax,color=['#4472C4','#70AD47']); ax.set_ylim(0,1.05); ax.set_ylabel('Score'); ax.set_title('Model Accuracy Comparison'); ax.legend(loc='lower right'); fig.tight_layout(); fig.savefig('images/05_model_accuracy_comparison.png',dpi=160,bbox_inches='tight'); plt.show()
```
