# Introduction

This notebook documents the **datasets, file formats, JSON structures,
and feature definitions** used throughout Project 5.

You will work with **two bankruptcy datasets**:

-   **Poland bankruptcy data** (features `feat_1` … `feat_64`)
-   **Taiwan bankruptcy data** (features `feat_1` … `feat_95`)

In both datasets, the prediction task is a **binary classification**
problem: predict whether a company will go bankrupt (`bankrupt = 1`) or
not (`bankrupt = 0`).

------------------------------------------------------------------------

# 1. Data files and where the records live

All files are stored in `data/` and are compressed JSON (`.json.gz`),
meaning:

-   the underlying format is **JSON**
-   the file is compressed using **gzip**
-   conceptually: “gzip wrapper” → “JSON text” → parsed into Python
    objects

There are **4 files**: 2 for Poland and 2 for Taiwan.

| Country | File | Contains target (`bankrupt`)? | Record list key |
|------|------------------------------------|-------------------:|-----------|
| Poland | `data/poland-bankruptcy-data-2009-mvp-features.json.gz` | No | `data` |
| Poland | `data/poland-bankruptcy-data-2009.json.gz` | Yes | `data` |
| Taiwan | `data/taiwan-bankruptcy-data-test-features.json.gz` | No | `observations` |
| Taiwan | `data/taiwan-bankruptcy-data.json.gz` | Yes | `observations` |

## 1.1 Top-level JSON layout

Each JSON file is a **top-level object** (a JSON “dictionary”) with a
few keys.

### 1.1.1 Poland JSON structure

Top-level keys:

-   `schema`
-   `data`
-   `metadata`

The actual records are stored under:

-   **`data`** → a list of company records

Each company record includes:

-   `company_id` (identifier)
-   `feat_1` … `feat_64`
-   optionally `bankrupt` (only in the “full” file)

### 1.1.1 Taiwan JSON structure

Top-level keys:

-   `schema`
-   `metadata`
-   `observations`

The actual records are stored under:

-   **`observations`** → a list of company records

Each company record includes:

-   `id` (identifier)
-   `feat_1` … `feat_95`
-   optionally `bankrupt` (only in the “full” file)

## 1.2. Why are there “features-only” files?

For each country, there is:

-   a **features-only** file (no `bankrupt`)
-   a **full** file (includes `bankrupt`)

This is a common real-world pattern:

-   you may receive features for pipeline development or testing
-   labels may arrive later, or be kept separate

In this project, you will use the “features-only” files for steps where
labels are not needed, and the “full” files when training/evaluating
models.

# 5. Practical reminders for later notebooks

## 5.1. Different feature counts

-   Poland: 64 features (`feat_1` … `feat_64`)
-   Taiwan: 95 features (`feat_1` … `feat_95`)

Code that validates “expected columns” must use the correct feature
range for each dataset.

## 5.2. Different record list keys

-   Poland records live in `payload["data"]`
-   Taiwan records live in `payload["observations"]`

This is why the wrangling code should not assume a single key name.

## 5.3. Features-only vs full datasets

-   features-only files do not include `bankrupt`
-   full files include `bankrupt`

When training/evaluating models, always confirm that `bankrupt` exists
in the DataFrame you are using.
