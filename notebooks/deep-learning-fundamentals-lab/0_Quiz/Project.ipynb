# 1. Linear algebra

## Essential knowledge

- Scalars are single numbers, vectors are one-dimensional, and matrices are two-dimensional.
- A dot product is `x @ y` in NumPy and equals the sum of pairwise products.
- An `(m, n)` matrix can multiply an `(n, p)` matrix, producing shape `(m, p)`.
- `*` is element-wise multiplication; `@` is matrix multiplication.
- Transposition swaps rows and columns, and `(AB)^T = B^T A^T`.
- A square matrix with determinant zero is singular and has no inverse.
- The Euclidean norm is the square root of the sum of squared components.

```python
# Shapes, dot product, norm, and matrix operations
x = np.array([1, 2, 3])
y = np.array([4, 0, -1])
A = np.array([[1, 2], [3, 4]])
B = np.array([[2, 0], [1, 2]])

print("x shape:", x.shape, "| A shape:", A.shape)
print("x dot y:", x @ y)
print("||x||:", np.linalg.norm(x))
print("A * B:\n", A * B)
print("A @ B:\n", A @ B)
print("B @ A:\n", B @ A)
print("transpose identity:", np.array_equal((A @ B).T, B.T @ A.T))
```

```python
# Determinant, inverse, and solving Ax = b
A = np.array([[2.0, 1.0], [1.0, 3.0]])
b = np.array([5.0, 7.0])
x = np.linalg.solve(A, b)

print("det(A):", np.linalg.det(A))
print("inverse(A):\n", np.linalg.inv(A))
print("solution:", x)
print("check Ax:", A @ x)
```

# 2. Calculus

## Essential knowledge

- A derivative is an instantaneous rate of change or tangent slope.
- Power rule: `d(x^n)/dx = n*x^(n-1)`.
- Product rule: `(fg)' = f'g + fg'`.
- Quotient rule: `(f/g)' = (f'g - fg')/g^2`.
- Chain rule: `d[f(g(x))]/dx = f'(g(x))*g'(x)`.
- The gradient contains all partial derivatives and points toward steepest increase.
- Gradient descent moves opposite the gradient: `theta_new = theta_old - rate * gradient`.

```python
# Numerical derivative of f(x)=x^2 at x=3
def numerical_derivative(f, x, h=1e-5):
    return (f(x + h) - f(x - h)) / (2 * h)

print("numerical:", numerical_derivative(lambda x: x**2, 3))
print("analytical:", 2 * 3)

# Gradient of f(x,y)=x^2+3y^2
def gradient(x, y):
    return np.array([2*x, 6*y])

print("gradient at (2,-1):", gradient(2, -1))
```

```python
# Gradient descent minimizes L(w)=(w-3)^2
def loss(w):
    return (w - 3)**2

w = -4.0
for step in range(20):
    grad = 2 * (w - 3)
    w = w - 0.1 * grad

print("w:", round(w, 4), "| loss:", round(loss(w), 6))
```

# 3. Probability and statistics

## Essential knowledge

- Complement: `P(not A)=1-P(A)`.
- Addition: `P(A or B)=P(A)+P(B)-P(A and B)`.
- Conditional probability: `P(A|B)=P(A and B)/P(B)`.
- Independence means `P(A and B)=P(A)P(B)`.
- Mutually exclusive events cannot occur together; they are usually not independent.
- Bayes: `P(A|B)=P(B|A)P(A)/P(B)`.
- The mean is sensitive to outliers; the median is more robust.
- Variance uses squared units; standard deviation uses original units.
- Correlation describes association, not causation.

```python
# Descriptive statistics and an outlier
data = [2, 3, 3, 4, 5]
with_outlier = data + [100]
for values in (data, with_outlier):
    print(values,
          "mean:", round(statistics.mean(values), 2),
          "median:", statistics.median(values),
          "population std:", round(statistics.pstdev(values), 2))
```

```python
# Probability rules and Bayes' theorem
p_heart = 13/52
p_face = 12/52
p_both = 3/52
print("P(heart or face):", p_heart + p_face - p_both)
print("P(face | heart):", p_both / p_heart)

prevalence, sensitivity, false_positive = 0.01, 0.90, 0.05
p_positive = sensitivity*prevalence + false_positive*(1-prevalence)
print("P(condition | positive):", round(sensitivity*prevalence/p_positive, 4))
```

```python
# Simulation versus exact binomial probability:
# exactly 3 heads in 5 fair tosses
rng = np.random.default_rng(42)
trials = rng.integers(0, 2, size=(100_000, 5))
simulated = np.mean(trials.sum(axis=1) == 3)
exact = math.comb(5, 3) * 0.5**5
print("simulated:", round(simulated, 4), "| exact:", exact)
```

## Question 1 - Singular matrices

**Question:** What does it mean for a matrix to be singular?

- All elements are the same
- It is not square
- **Its determinant equals zero (correct)**
- It has only one row

**Answer:** A square matrix is singular when its determinant is zero.

**Explanation:** A zero determinant means the matrix has no inverse. Its rows or columns are linearly dependent, so the corresponding linear transformation collapses at least one dimension. Consequently, a system Ax = b does not have a unique solution: depending on b, it has either no solution or infinitely many solutions.

The other choices do not define singularity. A non-square matrix has no ordinary determinant and is not normally classified as singular in this context. A one-row square matrix can be invertible if its only value is nonzero, and a matrix whose entries are all equal is singular only when its size is greater than 1 (or its single entry is zero).

```python
# Singular matrix example: the second row is twice the first
singular_matrix = np.array([[1.0, 2.0], [2.0, 4.0]])

determinant = np.linalg.det(singular_matrix)
rank = np.linalg.matrix_rank(singular_matrix)

print('determinant:', round(determinant, 10))
print('rank:', rank, '(less than 2, so the matrix is singular)')
assert np.isclose(determinant, 0.0)
```

## Question 2 - Matrix multiplication dimensions

**Question:** Why can't you multiply a 2 x 3 matrix by a 2 x 4 matrix?

- The matrices must be square
- **Inner dimensions don't match (3 != 2) (correct)**
- The first matrix must have more columns
- They have different total elements

**Answer:** The inner dimensions do not match: 3 is not equal to 2.

**Explanation:** For matrix multiplication AB, if A has shape (m, n), then B must have shape (n, p). The number of columns in A must equal the number of rows in B. Here, (2 x 3)(2 x 4) has inner dimensions 3 and 2, so the product is undefined.

A valid alternative is (2 x 3)(3 x 4), which produces a (2 x 4) matrix. Matrices do not need to be square, the first matrix does not simply need 'more columns,' and the total number of elements is irrelevant.

```python
# Matrix multiplication requires matching inner dimensions
A = np.ones((2, 3))
B_invalid = np.ones((2, 4))
B_valid = np.ones((3, 4))

print('Invalid shapes:', A.shape, '@', B_invalid.shape, 'because 3 != 2')
product = A @ B_valid
print('Valid shapes:', A.shape, '@', B_valid.shape, '->', product.shape)
assert product.shape == (2, 4)
```

## Question 3 - Dot product

**Question:** Calculate the dot product of [1, 2, 3] and [4, 5, 6].

- 21
- [4, 10, 18]
- 120
- **32 (correct)**

**Answer:** 32.

**Explanation:** Multiply corresponding components and then add the products:

(1 x 4) + (2 x 5) + (3 x 6) = 4 + 10 + 18 = 32.

A dot product returns a scalar. The vector [4, 10, 18] is only the element-wise product before summation. The other numerical choices do not result from the dot-product formula.

```python
# Dot product = sum of the element-wise products
u = np.array([1, 2, 3])
v = np.array([4, 5, 6])

elementwise_products = u * v
dot_product = u @ v

print('element-wise products:', elementwise_products)
print('dot product:', dot_product)
assert dot_product == 32
```

## Question 4 - Orthogonal vectors

**Question:** What does orthogonality mean for two vectors?

- They have the same length
- Their sum equals zero
- They point in the same direction
- **Their dot product equals zero (correct)**

**Answer:** Their dot product equals zero.

**Explanation:** Two vectors u and v are orthogonal when u dot v = 0. For two nonzero vectors, the dot-product identity u dot v = ||u|| ||v|| cos(theta) then implies cos(theta) = 0, so the angle theta between them is 90 degrees.

Orthogonal vectors do not need to have the same length. If their sum is zero, they point in opposite directions. Vectors pointing in the same direction generally have a positive, nonzero dot product.

```python
# The standard basis vectors are perpendicular
u = np.array([1, 0])
v = np.array([0, 1])

dot_product = u @ v
print('u dot v:', dot_product)
print('orthogonal:', np.isclose(dot_product, 0))
assert dot_product == 0
```

## Question 5 - Why a singular determinant is zero

**Question:** Why is the determinant of a singular matrix always zero?

- Singular matrices don't have determinants
- All singular matrices are zero matrices
- Definition of singular is zero determinant
- **Singular matrices have linearly dependent rows/columns (best explanation)**

**Answer:** Singular matrices have linearly dependent rows or columns.

**Explanation:** Linear dependence means that at least one row or column is a combination of the others. The matrix transformation therefore collapses space into a lower dimension, giving a zero area or volume scaling factor. Since the determinant measures this signed scaling factor, the determinant is zero.

For a square matrix, being singular, having determinant zero, having linearly dependent rows or columns, having rank below full rank, and having no inverse are equivalent conditions. Thus 'the definition of singular is zero determinant' identifies an equivalent criterion, but linear dependence provides the requested reason.

The other options are false: square singular matrices do have determinants, and a singular matrix need not be the zero matrix.

```python
# Row 2 is twice row 1, so the rows are linearly dependent
A = np.array([[1.0, 2.0], [2.0, 4.0]])

determinant = np.linalg.det(A)
rank = np.linalg.matrix_rank(A)

print('determinant:', round(determinant, 10))
print('rank:', rank, 'of', A.shape[0])
assert np.isclose(determinant, 0.0)
assert rank < A.shape[0]
```

## Question 6 - A 90-degree counterclockwise rotation

**Question:** Design a 2 x 2 matrix that rotates vectors by 90 degrees counterclockwise.

**Answer (third option):**

R=\begin{bmatrix}0 & -1 \\ 1 & 0\end{bmatrix}.

**Explanation:** A two-dimensional counterclockwise rotation through angle theta uses

R(\theta)=\begin{bmatrix}\cos\theta & -\sin\theta \\ \sin\theta & \cos\theta\end{bmatrix}.

For theta = 90 degrees, cos(theta) = 0 and sin(theta) = 1, producing the matrix above. It maps (1, 0) to (0, 1), which is a 90-degree counterclockwise turn.

The identity matrix leaves vectors unchanged; -I rotates them by 180 degrees; and [[0, 1], [-1, 0]] rotates them 90 degrees clockwise.

```python
# Rotate the unit x-vector 90 degrees counterclockwise
R_90_ccw = np.array([[0, -1], [1, 0]])
unit_x = np.array([1, 0])
rotated = R_90_ccw @ unit_x

print('rotation matrix:\n', R_90_ccw)
print('(1, 0) rotates to:', rotated)
assert np.array_equal(rotated, np.array([0, 1]))
assert np.allclose(R_90_ccw.T @ R_90_ccw, np.eye(2))
```

## Question 7 - Derivative using the power rule

**Question:** What is the derivative of f(x) = x^3?

- 3x^3
- **3x^2 (correct)**
- x^2
- 3x

**Answer:** f'(x) = 3x^2.

**Explanation:** The power rule states that d(x^n)/dx = n x^(n-1). For n = 3, bring the exponent down as a coefficient and reduce the exponent by one:

\frac{d}{dx}x^3 = 3x^{3-1} = 3x^2.

The exponent must decrease, so 3x^3 is incorrect. The coefficient 3 must also be retained, ruling out x^2 and 3x.

```python
# Compare the analytical derivative 3x^2 with a numerical estimate
f = lambda x: x**3
f_prime = lambda x: 3 * x**2
x = 2.0

numerical = numerical_derivative(f, x)
analytical = f_prime(x)

print('numerical derivative:', round(numerical, 6))
print('analytical derivative:', analytical)
assert np.isclose(numerical, analytical, atol=1e-4)
```

## Question 8 - Critical points

**Question:** What does a critical point of a function indicate?

- Points of discontinuity
- **Where the derivative equals zero or doesn't exist (correct)**
- Always a maximum or minimum
- Where the function equals zero

**Answer:** A critical point occurs at a point in the function's domain where the derivative is zero or does not exist.

**Explanation:** Critical points are candidates for local maxima, local minima, or neither. Further analysis, such as checking derivative sign changes or using the second-derivative test, is required to classify them.

For example, f(x) = x^3 has f'(0) = 0, but x = 0 is neither a maximum nor a minimum; it is a stationary inflection point. A root is where f(x) = 0, which is a different concept. A discontinuity is not automatically a critical point because the point must belong to the function's domain.

```python
# f(x)=x^3 has a critical point at zero but no extremum there
f = lambda x: x**3
f_prime = lambda x: 3 * x**2

critical_x = 0.0
print('f prime at zero:', f_prime(critical_x))
print('nearby values:', f(-0.1), f(critical_x), f(0.1))
print('The function keeps increasing through zero.')
assert f_prime(critical_x) == 0
assert f(-0.1) < f(critical_x) < f(0.1)
```

## Question 9 - Product rule

**Question:** Apply the product rule to find the derivative of x times e^x.

- e^x
- **e^x + x e^x (correct)**
- x + e^x
- x e^x

**Answer:**

\frac{d}{dx}(x e^x)=e^x+x e^x.

**Explanation:** The product rule is (uv)' = u'v + uv'. Let u = x and v = e^x. Then u' = 1 and v' = e^x, so

(x e^x)'=(1)(e^x)+x(e^x)=e^x+x e^x=e^x(1+x).

Differentiating only one factor produces e^x or x e^x and misses half of the product rule. Adding the original factors as x + e^x is not the product rule.

```python
# Verify d(x*e^x)/dx = e^x + x*e^x numerically
f = lambda x: x * np.exp(x)
f_prime = lambda x: np.exp(x) + x * np.exp(x)
x = 1.5

numerical = numerical_derivative(f, x)
analytical = f_prime(x)

print('numerical derivative:', round(numerical, 6))
print('analytical derivative:', round(analytical, 6))
assert np.isclose(numerical, analytical, atol=1e-4)
```

## Question 10 - Chain rule

**Question:** Find the derivative of f(x) = sin(2x).

- 2 sin(2x)
- **2 cos(2x) (correct)**
- -2 cos(2x)
- cos(2x)

**Answer:** f'(x) = 2 cos(2x).

**Explanation:** Use the chain rule. Let u = 2x. The derivative of sin(u) is cos(u), and du/dx = 2. Therefore,

\frac{d}{dx}\sin(2x)=\cos(2x)\cdot 2=2\cos(2x).

The derivative of sine is positive cosine, so the negative option is incorrect. The factor 2 comes from differentiating the inner function 2x; omitting it gives cos(2x).

```python
# Verify d[sin(2x)]/dx = 2*cos(2x) numerically
f = lambda x: np.sin(2 * x)
f_prime = lambda x: 2 * np.cos(2 * x)
x = 0.7

numerical = numerical_derivative(f, x)
analytical = f_prime(x)

print('numerical derivative:', round(numerical, 6))
print('analytical derivative:', round(analytical, 6))
assert np.isclose(numerical, analytical, atol=1e-4)
```

## Question 11 - Continuity and integrability

**Question:** What's the relationship between continuity and integrability?

- They're independent properties
- **Continuous functions are always integrable (correct in the standard quiz context)**
- Neither implies the other
- Integrable functions are always continuous

**Answer:** A continuous function on a closed, bounded interval is integrable on that interval.

**Explanation:** Continuity is a sufficient condition for Riemann integrability on [a, b]. A continuous function on such an interval is bounded and has no discontinuities, so its definite integral exists.

The converse is false: an integrable function need not be continuous. For example, a step function with a jump at one point is still Riemann integrable. Thus continuity implies integrability under the standard closed-and-bounded-interval assumptions, but integrability does not imply continuity.

**Qualification:** On unbounded intervals or for functions with singular endpoints, continuity alone does not guarantee that an improper integral converges. The quiz answer assumes ordinary integration on a finite closed interval.

```python
# A discontinuous step function is still integrable on [0, 1]
def step_function(x):
    return np.where(x < 0.5, 0.0, 1.0)

x = np.linspace(0, 1, 100_001)
area = np.trapezoid(step_function(x), x)

print('approximate integral:', round(area, 6))
print('The function jumps at x=0.5 but its integral is 0.5.')
assert np.isclose(area, 0.5, atol=1e-4)
```

## Question 12 - Designing a function with a given tangent slope

**Question:** Design a function where the tangent line at x = 1 has slope 3.

- f(x) = (3x^2)/2 + c
- f(x) = x + 2
- f(x) = x^3
- f(x) = 3x

**Answer:** The question has multiple mathematically correct choices. The simplest, and likely intended, answer is **f(x) = 3x**. However, f(x) = (3x^2)/2 + c and f(x) = x^3 also satisfy the condition.

**Explanation:** The tangent slope at x = 1 is f'(1), so test every option:

- If f(x) = (3x^2)/2 + c, then f'(x) = 3x and f'(1) = 3.
- If f(x) = x + 2, then f'(x) = 1, so it does not qualify.
- If f(x) = x^3, then f'(x) = 3x^2 and f'(1) = 3.
- If f(x) = 3x, then f'(x) = 3 and f'(1) = 3.

Therefore options 1, 3, and 4 are valid. If the interface requires exactly one selection, f(x) = 3x is the most direct construction, but the single-answer question is not mathematically well-posed.

```python
# Evaluate each option's derivative at x=1
derivatives_at_1 = {
    '(3x^2)/2 + c': 3 * 1,
    'x + 2': 1,
    'x^3': 3 * 1**2,
    '3x': 3,
}

valid_options = [name for name, slope in derivatives_at_1.items() if slope == 3]
print('derivatives at x=1:', derivatives_at_1)
print('valid choices:', valid_options)
assert valid_options == ['(3x^2)/2 + c', 'x^3', '3x']
```

**Answer:** The population is the complete group of interest, while a sample is a subset selected from that population.

**Explanation:** Researchers often study a sample because measuring every member of a population is expensive or impractical. A numerical property of a population, such as its true mean, is called a parameter. The corresponding value calculated from a sample is a statistic and is used to estimate the parameter.

A population can be real or conceptual; it is not necessarily theoretical. A sample is normally smaller than its population, and the two terms are not interchangeable. A representative sampling method is important because a biased sample can produce misleading estimates.

```python
# A sample estimates a property of the full population
population = np.arange(1, 101)
rng = np.random.default_rng(13)
sample = rng.choice(population, size=10, replace=False)

print('population size:', population.size)
print('sample size:', sample.size)
print('population mean (parameter):', population.mean())
print('sample mean (statistic):', sample.mean())
assert sample.size < population.size
assert set(sample).issubset(set(population))
```

## Question 14 - Correlation

**Question:** What does correlation measure?

- The sum of variables
- **Linear relationship between variables (correct)**
- Variable independence
- Causation between variables

**Answer:** Correlation measures the strength and direction of a linear relationship between two variables.

**Explanation:** The Pearson correlation coefficient ranges from -1 to 1. Values near 1 indicate a strong positive linear relationship, values near -1 indicate a strong negative linear relationship, and values near 0 indicate little or no linear relationship.

Correlation does not establish causation. Also, zero correlation does not necessarily imply independence: variables can have a strong nonlinear relationship while their linear correlation is zero. Independence does imply zero correlation when the relevant moments exist, but the converse is generally false.

```python
# Pearson correlation captures linear association
x = np.array([-2, -1, 0, 1, 2], dtype=float)
positive = 3 * x + 1
negative = -2 * x + 4
nonlinear = x**2

corr_positive = np.corrcoef(x, positive)[0, 1]
corr_negative = np.corrcoef(x, negative)[0, 1]
corr_nonlinear = np.corrcoef(x, nonlinear)[0, 1]

print('positive linear correlation:', round(corr_positive, 3))
print('negative linear correlation:', round(corr_negative, 3))
print('x versus x^2 correlation:', round(corr_nonlinear, 3))
assert np.isclose(corr_positive, 1.0)
assert np.isclose(corr_negative, -1.0)
assert np.isclose(corr_nonlinear, 0.0)
```

## Question 15 - Probability of a dice sum

**Question:** What's the probability of rolling a sum of 7 with two dice?

- 1/7
- **1/6 (correct)**
- 7/36
- 1/12

**Answer:** 1/6.

**Explanation:** Two fair six-sided dice have 6 x 6 = 36 equally likely ordered outcomes. Six outcomes sum to 7:

(1,6), (2,5), (3,4), (4,3), (5,2), and (6,1).

Therefore,

P(\text{sum}=7)=\frac{6}{36}=\frac{1}{6}.

The numerator counts successful ordered pairs, not the target sum itself, so 7/36 is incorrect.

```python
# Enumerate all ordered outcomes for two fair six-sided dice
outcomes = [(die_1, die_2) for die_1 in range(1, 7) for die_2 in range(1, 7)]
sum_of_seven = [outcome for outcome in outcomes if sum(outcome) == 7]
probability = len(sum_of_seven) / len(outcomes)

print('successful outcomes:', sum_of_seven)
print('count:', len(sum_of_seven), 'out of', len(outcomes))
print('probability:', probability)
assert len(sum_of_seven) == 6
assert np.isclose(probability, 1 / 6)
```

## Question 17 - Central Limit Theorem

**Question:** How does the Central Limit Theorem work?

- Limits sample size
- All data becomes normal
- **Sample means become normal as n increases (correct)**
- Centers data at zero

**Answer:** The sampling distribution of the sample mean becomes approximately normal as the sample size n increases.

**Explanation:** For independent, identically distributed observations with finite mean mu and finite variance sigma squared, the distribution of sample means approaches a normal distribution as n grows, even when the population itself is not normal. The sampling distribution is centered at mu and has standard error sigma divided by the square root of n.

The theorem does not make the original observations normal, does not force the mean to zero, and does not limit sample size. It describes the distribution obtained from the means of many repeated samples. How large n must be depends on the population's skewness and tail behavior.

```python
# Exponential data are skewed, but means of large samples are more normal-shaped
rng = np.random.default_rng(17)
sample_size = 40
repetitions = 20_000
sample_means = rng.exponential(scale=1.0, size=(repetitions, sample_size)).mean(axis=1)

print('mean of sample means:', round(sample_means.mean(), 3))
print('std of sample means:', round(sample_means.std(), 3))
print('theoretical standard error:', round(1 / np.sqrt(sample_size), 3))
assert np.isclose(sample_means.mean(), 1.0, atol=0.02)
assert np.isclose(sample_means.std(), 1 / np.sqrt(sample_size), atol=0.01)
```

## Question 20 - Function-local scope

**Question:** What's the scope of a variable defined inside a function?

- Available in all functions
- Global everywhere
- Deleted after one use
- **Local to that function (correct)**

**Answer:** A variable assigned inside a function is normally local to that function.

**Explanation:** A local variable can be referenced inside the function call in which it is created, but it is not directly available from unrelated functions or global code. It is not deleted after one use; it can be used repeatedly while that function call is active.

Python resolves names using the LEGB order: Local, Enclosing, Global, then Built-in scope. The global keyword can make an assignment target a global name, and nonlocal can target a name in an enclosing function, but these are explicit exceptions to normal local assignment.

```python
# local_value exists only inside make_value
def make_value():
    local_value = 42
    return local_value

result = make_value()
print('returned value:', result)
print('local_value exists globally:', 'local_value' in globals())
assert result == 42
assert 'local_value' not in globals()
```

**Question:** What does x is y check in Python?

**Answer:** The is operator checks whether x and y refer to the exact same object.

## Question 22 - Mixing strings and integers

**Question:** Why does "2" + 3 raise an error?

- Strings cannot contain numbers
- **Python does not implicitly convert types (correct)**
- 3 is not a string
- Addition is undefined

**Answer:** Python does not implicitly convert the string and integer to a common type for the + operation.

**Explanation:** The + operator performs numeric addition between compatible numbers and concatenation between strings. Here, "2" is a str and 3 is an int. Python requires an explicit choice between arithmetic and concatenation, so the expression raises TypeError.

For arithmetic, int("2") + 3 gives 5. For concatenation, "2" + str(3) gives "23". Strings can contain digit characters, and addition itself is well-defined when the operands have compatible types. Although 3 is indeed not a string, the more general reason is the lack of implicit conversion here.

```python
# Choose the intended operation by converting explicitly
numeric_result = int('2') + 3
text_result = '2' + str(3)

print('numeric addition:', numeric_result)
print('string concatenation:', text_result)

try:
    invalid_result = '2' + 3
except TypeError as error:
    print('mixed types raise:', type(error).__name__)

assert numeric_result == 5
assert text_result == '23'
```

## Question 23 - self in instance methods

**Question:** What's the purpose of self in class methods?

- **References the current instance (correct)**
- Defines class variables
- Creates new instances
- Optional parameter

**Answer:** self references the current object instance on which an instance method is being called.

**Explanation:** Instance methods conventionally name their first parameter self. Python automatically passes the instance for a bound-method call, allowing the method to read or modify instance attributes such as self.name. The parameter is required in the method definition, although the caller does not pass it explicitly.

self does not create instances; the class constructor call does that. Assignments on the class itself define class variables, while assignments such as self.value define or update instance attributes. Technically, self is a naming convention rather than a reserved keyword, but following the convention is essential for readable Python.

**Terminology note:** Despite the wording of the question, this describes instance methods. A method decorated with @classmethod conventionally receives cls, which references the class.

```python
class Counter:
    def __init__(self, start=0):
        self.value = start

    def increment(self):
        self.value += 1
        return self.value

counter = Counter(10)
bound_result = counter.increment()
# Equivalent explicit call: Counter.increment(counter)
explicit_result = Counter.increment(counter)

print('after bound call:', bound_result)
print('after explicit call:', explicit_result)
assert counter.value == 12
```

## Question 24 - Python type hints

**Question:** Evaluate: "Type hints make Python statically typed." Is this correct?

- **False. Type hints are optional and not enforced (correct)**
- Depends on the interpreter
- True. Types are checked at compile time
- True for Python 3.5+

**Answer:** False. Python remains dynamically typed; type hints are optional annotations and are not ordinarily enforced at runtime.

**Explanation:** Type hints communicate intended types to readers, IDEs, linters, and optional static type checkers such as mypy or Pyright. Those tools can report mismatches before execution, but the standard Python runtime usually does not reject a function call merely because an argument conflicts with an annotation.

Annotations can be inspected through a function's __annotations__ attribute. Separate libraries or explicit runtime checks can enforce them, but that behavior is not provided automatically by normal type hints. Python 3.5 introduced the modern type-hinting system, but its presence did not turn Python into a statically typed language.

```python
# Type annotations describe intent but are not automatically enforced
def repeat_text(text: str, times: int) -> str:
    return text * times

print('annotations:', repeat_text.__annotations__)
print('normal annotated call:', repeat_text('ha', 3))

# bool is not the intended int argument here, but normal runtime execution
# still permits the call because bool supports the required operation.
runtime_result = repeat_text('ha', True)
print('runtime accepted mismatched annotation:', runtime_result)
assert runtime_result == 'ha'
```
